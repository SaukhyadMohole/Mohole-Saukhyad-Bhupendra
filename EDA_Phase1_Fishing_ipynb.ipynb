```python
df = df.drop_duplicates()
```

```python
df.dtypes
```

```python
df.columns = df.columns.str.strip()

df.columns
```

```python
print("Missing values in dataset:", df.isnull().sum().sum())
```

```python
num_cols = df.select_dtypes(include=["number"]).columns

df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
cat_cols = df.select_dtypes(include=["object"]).columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["mode"] = encoder.fit_transform(df["mode"])

df.head()
```

```python
mapping = dict(zip(encoder.classes_, encoder.transform(encoder.classes_)))

print(mapping)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

num_cols = [
    "rownames",
    "price",
    "catch",
    "pbeach",
    "ppier",
    "pboat",
    "pcharter",
    "cbeach",
    "cpier",
    "cboat",
    "ccharter",
    "income"
]

df[num_cols] = scaler.fit_transform(df[num_cols])

df.head()
```

```python
df.describe().round(2)
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["price"], bins=20, kde=True)

plt.title("Distribution of Price")
plt.xlabel("Price")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["catch"], bins=20, kde=True)

plt.title("Distribution of Catch")
plt.xlabel("Catch")
plt.ylabel("Frequency")

plt.show()
```

```python
mode_names = encoder.inverse_transform(df["mode"])

plt.figure(figsize=(8,5))
sns.countplot(x=mode_names)

plt.title("Fishing Mode Distribution")
plt.xlabel("Mode")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["income"])

plt.title("Boxplot of Income")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="price", y="catch", data=df)

plt.title("Price vs Catch")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="income", y="catch", data=df)

plt.title("Income vs Catch")

plt.show()
```

```python
plot_df = df.copy()

plot_df["mode"] = encoder.inverse_transform(plot_df["mode"])

plt.figure(figsize=(8,5))
sns.boxplot(x="mode", y="price", data=plot_df)

plt.title("Price by Fishing Mode")
plt.xlabel("Mode")
plt.ylabel("Price")

plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.pairplot(df[["price", "catch", "income", "mode"]])

plt.show()
```

```python
plot_df = df.copy()

plot_df["mode"] = encoder.inverse_transform(plot_df["mode"])

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=plot_df,
    x="price",
    y="catch",
    hue="mode"
)

plt.title("Price vs Catch by Fishing Mode")

plt.show()
```
